# **Razonamiento**

Los LLMs (Modelos de Lenguaje Grandes) se han vuelto increíblemente buenos en tareas de generación de texto. Pueden escribir ensayos, resumir documentos e incluso mantener conversaciones complejas. Sin embargo, cuando se trata de razonar y resolver problemas complejos de múltiples pasos, a menudo fallan.

El objetivo del razonamiento en los LLMs es lograr que generen una secuencia lógica de pasos para alcanzar la respuesta correcta, en lugar de saltar directamente a una respuesta potencialmente incorrecta.

<div align="center">
  <img src="imgs/reasoning02.jpeg" width="500px">
</div>

Hay dos formas principales de mejorar el razonamiento de un LLM:

- **Activación de Razonamiento:** Técnicas de prompting (o decoding) que fuerzan a un modelo ya existente a razonar mejor.

- **Entrenamiento para Razonar:** Entrenamiento especializado que enseña al modelo a priorizar la generación de tokens que reflejan un razonamiento lógico.

## **Activación de Razonamiento**

### **1. Chain-of-Thought Prompting (CoT)**

Mecanismos de Prompting que fomentan el razonamiento paso a paso. Para problemas complejos, los LLMs tienden a tener dificultades para llegar a la respuesta correcta en un solo paso. CoT ayuda al modelo a descomponer el problema en pasos intermedios. Así, el modelo puede razonar a través de cada paso antes de llegar a la respuesta final.

##### **<div style="color:navy;">1.1.  CoT Few-Shot</div>** 

Consiste en proporcionar al modelo ejemplos de preguntas junto con sus razonamientos paso a paso antes de hacer la pregunta objetivo. Por ejemplo:


<div style="background-color: #e4f0f8ff; border-left: 5px solid #ffffffff; padding: 1.5em; margin: 30px; width: 600px">
<code>
EJEMPLO 1:

Pregunta: En una caja hay 24 lápices. Si 1/4 de ellos son azules y el resto son rojos, ¿cuántos lápices rojos hay?

Razonamiento:
1. El número total de lápices es 24.
2. Primero, calculamos cuántos lápices son azules: 1/4 de 24 es 24 / 4 = 6 lápices azules.
3. Luego, restamos los azules del total para encontrar los rojos: 24 - 6 = 18 lápices rojos.

Respuesta: 18

----------------------------------------------------------------------

PREGUNTA 2: Una escuela tiene 90 estudiantes. Si 2/3 de los estudiantes llevan mochila y el resto no, ¿cuántos estudiantes NO llevan mochila?

Razonamiento:
</code>
</div>




##### **<div style="color:navy;">1.2.  CoT Zero-Shot</div>** 
En lugar de proporcionar ejemplos, simplemente se añade una indicación al final de la pregunta que invita al modelo a razonar paso a paso, como "Let's think step by step." Por ejemplo:



<div style="background-color: #e4f0f8ff; border-left: 5px solid #ffffffff; padding: 1.5em; margin: 30px; width: 600px">
<code>
Q: Si tengo 7 manzanas, doy 3 a mi amigo, luego me como otra y compro 5 más. ¿Cuántas manzanas tengo ahora? Razona paso a paso.
<br><br>
A: Primero, empiezo con 7 manzanas. Luego, doy 3 a mi amigo, así que me quedan 4 manzanas. Después, me como 1 manzana, quedándome con 3. Finalmente, compro 5 más, lo que me da un total de 8 manzanas.
</code>
</div>






### **2. Autoconsistencia (Self-Consistency)**

Mecanismo de Muestreo: Generar múltiples cadenas de pensamiento ($N$).
Selección: Votación por mayoría para elegir la respuesta más frecuente.

### **3. Búsqueda Estructurada y Exploratoria**

Árbol de Pensamiento (Tree-of-Thought, ToT): Razonamiento que se ramifica y utiliza una evaluación heurística de estados (búsqueda).

Grafo de Pensamiento (Graph-of-Thought, GoT): Conexiones no lineales entre los pasos de razonamiento (nodos).

### **4. Auto-Corrección (Self-Correction/Reflection)**

El modelo genera una solución y luego actúa como crítico para refinarla.

¡Excelente pregunta! La clasificación de las técnicas de mejora del razonamiento en LLMs según el momento en que se aplican (**tiempo de entrenamiento** vs. **tiempo de prueba/inferencia**) es fundamental para la ingeniería de sistemas.

Aquí tienes un desglose de las técnicas de razonamiento en LLMs clasificadas como "Train-Time" y "Test-Time":

---

## ⏳ Técnicas de Razonamiento "Test-Time" (Tiempo de Prueba / Inferencia)

Estas técnicas se aplican **después** de que el LLM ha sido completamente entrenado (o preentrenado). No modifican los pesos o parámetros del modelo; en su lugar, manipulan la **entrada (*prompt*)** o el **proceso de generación (*decoding*)** durante la inferencia para activar el razonamiento.

| Técnica | Descripción | Mecanismo |
| :--- | :--- | :--- |
| **Chain-of-Thought (CoT)** | La técnica básica para guiar el modelo a generar pasos intermedios. | Modificación del *prompt* (instrucción o ejemplos). |
| **CoT *Zero-Shot*** | Añadir la frase "Pensemos paso a paso" (*"Let's think step by step"*). | Manipulación del *prompt* para activar el razonamiento latente. |
| **Autoconsistencia (*Self-Consistency*)** | Generar $N$ respuestas con CoT y elegir la respuesta por voto mayoritario. | Manipulación del **proceso de *decoding*** (muestreo) y agregación. |
| **Árbol de Pensamiento (ToT)** | El modelo explora **múltiples ramas** de razonamiento y utiliza un algoritmo de búsqueda para elegir el camino más prometedor. | Manipulación del **proceso de *decoding*** (búsqueda estructurada). |
| **Auto-Corrección / Reflexión** | El modelo genera una crítica de su propia respuesta y la corrige. | **Múltiples pasadas** del *prompt* (generación e inserción de *feedback*). |
| **Tool-Use / Function Calling** | El modelo genera un JSON para llamar a una herramienta externa (ej., Python o API) que es ejecutada por un **Agente Orquestador**. | Manipulación de la **salida del modelo** y uso de **software externo** (TTC). |

---

## 🏋️ Técnicas de Razonamiento "Train-Time" (Tiempo de Entrenamiento)

Estas técnicas se aplican **durante** o **inmediatamente después** de la fase de preentrenamiento. Modifican directamente los pesos del modelo para mejorar su capacidad nativa de generar razonamiento correcto y lógico.

| Técnica | Descripción | Objetivo en el Modelo |
| :--- | :--- | :--- |
| **Ajuste Fino Supervisado (SFT) con CoT** | Entrenar el modelo con un *dataset* extenso de problemas emparejados con **cadenas de razonamiento completas**. | Enseñar al modelo a **generar el formato y la estructura del razonamiento** de forma nativa. |
| **Aprendizaje por Refuerzo (RL)** | Usar RL para optimizar la política de generación del modelo. | Optimizar los pesos para que el modelo priorice la **generación de *tokens* que conducen a la respuesta correcta** (razonamiento lógico). |
| **GRPO (Group Relative Policy Optimization)** | Algoritmo de RL que utiliza un **Verificador Programado** para recompensar la corrección y ejecutabilidad del razonamiento. | Ajustar los pesos para maximizar la **precisión determinística** del razonamiento (ej., en matemáticas y código). |
| **Entrenamiento de Llamada a Función (Tool-Calling)** | Entrenar el modelo con ejemplos que muestren cuándo y cómo generar la sintaxis de llamada a función (ej., el JSON). | Enseñar al modelo a **reconocer la necesidad de la herramienta** y generar el formato de salida requerido. |

---

### Resumen y Conexión 🔗

Las técnicas más avanzadas para LLMs, como las utilizadas en modelos de vanguardia, a menudo combinan ambos enfoques:

1.  **Train-Time:** Se utiliza SFT y RL (como GRPO) para crear un **modelo base con una fuerte capacidad nativa de razonamiento**.
2.  **Test-Time:** Se utiliza **Tool-Use** y **ToT/Self-Consistency** en la inferencia para desbloquear todo el potencial de razonamiento y garantizar la precisión final.
