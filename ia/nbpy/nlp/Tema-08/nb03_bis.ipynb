# **Guía Visual sobre el Razonamiento en LLMs**
### **Explorando las Técnicas de Cómputo en Tiempo de Prueba y DeepSeek-R1**


## **¿Qué son los LLMs de Razonamiento?**

En comparación con los LLMs regulares, los **LLMs de razonamiento** tienden a **descomponer un problema en pasos más pequeños** (a menudo llamados pasos de razonamiento o procesos de pensamiento) antes de responder una pregunta dada.

<div align="center">
  <img src="imgs/reasoning01.jpeg" width="500px">
</div>

Entonces, ¿qué significa realmente un "proceso de pensamiento", un "paso de razonamiento" o una "Cadena de Pensamiento"? Aunque podemos filosofar si los LLMs son realmente capaces de pensar como los humanos, estos pasos de razonamiento dividen el proceso en **inferencias más pequeñas y estructuradas**.

<div align="center">
  <img src="imgs/reasoning02.jpeg" width="500px">
</div>

En otras palabras, en lugar de que los LLMs aprendan "qué" responder, ¡aprenden "cómo" responder! Para entender cómo se crean los LLMs de razonamiento, primero exploramos el cambio de paradigma con un enfoque en el escalado durante el entrenamiento (**cómputo en tiempo de entrenamiento**) a un enfoque en la inferencia (**cómputo en tiempo de prueba**).


## **¿Qué es el Cómputo en Tiempo de Entrenamiento?**

Hasta la mitad de 2024, para aumentar el rendimiento de los LLMs durante el **preentrenamiento**, los desarrolladores a menudo aumentaban el **tamaño** de:

* El Modelo (número de **parámetros**)
* El Conjunto de Datos (número de **tokens**)
* El Cómputo (número de **FLOPs**)

En conjunto, esto se llama **cómputo en tiempo de entrenamiento,** que se refiere a la idea de que los datos de preentrenamiento son el "combustible fósil de la IA". Esencialmente, cuanto mayor sea su presupuesto de preentrenamiento, mejor será el modelo resultante.

<div align="center">
  <img src="imgs/reasoning03.jpeg" width="500px">
</div>

El cómputo en tiempo de entrenamiento puede incluir tanto el cómputo necesario durante el entrenamiento como durante el ajuste fino (*fine-tuning*).

<div align="center">
  <img src="imgs/reasoning04.jpeg" width="500px">
</div>

Juntos, han sido un enfoque principal para aumentar el rendimiento de los LLMs.

### **Leyes de escalado**

La forma en que la escala de un modelo (a través del cómputo, el tamaño del conjunto de datos y el tamaño del modelo) se correlaciona con el rendimiento de un modelo se investiga a través de varias leyes de escalado. Son las llamadas "**Power Laws**" donde un aumento en una variable (ej., cómputo) resulta en un incremento proporcional en otra variable (ej., rendimiento).

<div align="center">
  <img src="imgs/reasoning05.jpeg" width="500px">
</div>

Las más conocidas son las leyes de escalado de **"Kaplan"** y **"Chinchilla"**. Estas leyes dicen más o menos que el rendimiento de un modelo aumentará con más cómputo, *tokens* y parámetros.

<div align="center">
  <img src="imgs/reasoning06.jpeg" width="500px">
</div>

Estas gráficas sugieren que los tres factores deben escalarse conjuntamente para un rendimiento óptimo.

La **Ley de escalado de Kaplan** establece que escalar el tamaño del modelo es típicamente más efectivo que escalar los datos (dado un cómputo fijo). En contraste, la **Ley de escalado de Chinchilla** sugiere que el tamaño del modelo y los datos son igual de importantes.

Sin embargo, a lo largo de 2024, el cómputo, el tamaño del conjunto de datos y los parámetros del modelo han crecido constantemente, pero las ganancias han mostrado rendimientos decrecientes.

<div align="center">
  <img src="imgs/reasoning07.jpeg" width="500px">
</div>

Hay rendimientos decrecientes a medida que se escala. Esto plantea la pregunta: "¿Hemos llegado a un muro?"


## **¿Qué es el Cómputo en Tiempo de Test?**

La naturaleza costosa de aumentar el cómputo en tiempo de entrenamiento llevó a un interés en un enfoque alternativo, el **cómputo en tiempo de test**. En lugar de aumentar continuamente los presupuestos de preentrenamiento, el cómputo en tiempo de test permite que los modelos "*piensen más tiempo*" durante la **inferencia**.

<div align="center">
  <img src="imgs/reasoning08.jpeg" width="500px">
</div>

Con los modelos que no razonan, normalmente solo se obtendría la respuesta y se omitirían los pasos de "razonamiento":

<div align="center">
  <img src="imgs/reasoning09.jpeg" width="500px">
</div>

Sin embargo, los modelos de razonamiento usarían más *tokens* para derivar su respuesta a través de un proceso de **"pensamiento"** sistemático:

<div align="center">
  <img src="imgs/reasoning10.jpeg" width="500px">
</div>

La idea es que el LLM tiene que gastar recursos (como el cómputo de VRAM) para crear una respuesta. Sin embargo, si todo el cómputo se gasta generando la respuesta, ¡eso es un poco ineficiente! En cambio, al crear más *tokens* de antemano que contienen información adicional, relaciones y nuevos pensamientos, el modelo gasta más cómputo generando la respuesta final.

<div align="center">
  <img src="imgs/reasoning11.jpeg" width="500px">
</div>

## **Leyes de escalado para el Cómputo en Tiempo de Test**

En comparación con el cómputo en tiempo de entrenamiento, las leyes de escalado para el cómputo en tiempo de prueba son relativamente nuevas. Cabe destacar dos fuentes interesantes que relacionan el escalado del cómputo en tiempo de prueba con el cómputo en tiempo de entrenamiento.

Primero, una [publicación de OpenAI](https://openai.com/index/learning-to-reason-with-llms/) muestra que el cómputo en tiempo de prueba podría seguir la misma tendencia que el escalado del cómputo en tiempo de entrenamiento.

<div align="center">
  <img src="imgs/reasoning12.jpeg" width="500px">
</div>

Como tal, afirman que podría haber un cambio de paradigma hacia el escalado del cómputo en tiempo de prueba, ya que todavía es un campo nuevo.

Segundo, un interesante artículo llamado "Scaling Scaling Laws with Board Games" explora **AlphaZero** y lo entrena con varios grados de cómputo para jugar a Hex.

[Image: Figura anotada de "Scaling Scaling Laws with Board Games". Muestra la construcción de cómputo en tiempo de entrenamiento y cómputo en tiempo de prueba variables.]

Sus resultados muestran que el cómputo en tiempo de entrenamiento y el cómputo en tiempo de prueba están estrechamente relacionados. Cada línea punteada muestra el cómputo mínimo necesario para esa puntuación ELO particular.

[Image: Figura anotada de "Scaling Scaling Laws with Board Games". Demuestra la relación entre el cómputo en tiempo de entrenamiento y el cómputo en tiempo de prueba.]

Con el cómputo en tiempo de prueba escalando como el cómputo en tiempo de entrenamiento, se está produciendo un cambio de paradigma hacia modelos de "razonamiento" que utilizan más cómputo en tiempo de prueba.

A través de este cambio de paradigma, en lugar de centrarse puramente en el cómputo en tiempo de entrenamiento (preentrenamiento y ajuste fino), estos modelos de "razonamiento" en cambio equilibran el entrenamiento con la inferencia.

[Image: Diagrama que muestra el equilibrio entre el Cómputo en Tiempo de Entrenamiento y el Cómputo en Tiempo de Prueba.]

## **Categorías de Cómputo en Tiempo de Test**

El increíble éxito de modelos de razonamiento como DeepSeek R-1 y OpenAI o1 muestran que hay más técnicas que simplemente pensar "más tiempo".

Como exploraremos, el cómputo en tiempo de prueba puede ser muchas cosas diferentes, incluyendo Cadena de Pensamiento, revisión de respuestas, retroceso, muestreo y más.

Estos se pueden clasificar, a grandes rasgos, en dos categorías:

* **Búsqueda contra Verificadores** (*Search against Verifiers*) (muestrear generaciones y seleccionar la mejor respuesta)
* **Modificación de la Distribución de Propuesta** (*Modifying Proposal Distribution*) (proceso de "pensamiento" entrenado)

[Image: Diagrama que muestra las dos categorías de técnicas de razonamiento.]

Por lo tanto, la búsqueda contra verificadores está **centrada en la salida** (*output*-focused), mientras que la modificación de la distribución de propuesta está **centrada en la entrada** (*input*-focused).

[Image: Diagrama que compara los enfoques centrado en la salida (Search) y centrado en la entrada (Modifying Proposal Distribution).]

Hay dos tipos de verificadores que exploraremos:

* Modelos de Recompensa de Resultado (*Outcome Reward Models*, **ORM**)
* Modelos de Recompensa de Proceso (*Process Reward Models*, **PRM**)

Como sus nombres indican, el **ORM** solo juzga el resultado y no se preocupa por el proceso subyacente:

[Image: Diagrama que ilustra un ORM juzgando solo el resultado.]

En contraste, el **PRM** también juzga el proceso que conduce al resultado (el "razonamiento"):

[Image: Diagrama que ilustra un PRM juzgando cada paso del razonamiento.]

Observe cómo el paso 2 es un paso de razonamiento deficiente y es puntuado bajo por el PRM!

Ahora que tiene una buena comprensión de los ORM frente a los PRM, ¡exploremos cómo se pueden aplicar en varias técnicas de verificación!

---

## Búsqueda contra Verificadores

La primera categoría principal de cómputo en tiempo de prueba es la **búsqueda contra verificadores**. Esto generalmente implica dos pasos.

* Primero, se crean **múltiples muestras** de procesos de razonamiento y respuestas.
* Segundo, un verificador (*Reward Model*) puntúa la salida generada

[Image: Diagrama que ilustra la creación de múltiples muestras de respuesta.]

El verificador es típicamente un LLM, ajustado para juzgar el resultado (ORM) o el proceso (PRM).

Una ventaja importante de usar verificadores es que **no es necesario volver a entrenar o ajustar fino el LLM** que se utiliza para responder a la pregunta.

### Votación por Mayoría (*Majority Voting*)

El método más sencillo es en realidad no utilizar un modelo de recompensa o verificador, sino realizar una **votación por mayoría**.

Permitimos que el modelo genere múltiples respuestas y la respuesta que se genera con mayor frecuencia será la respuesta final.

[Image: Diagrama que ilustra la Votación por Mayoría (Autoconsistencia).]

Este método también se llama **autoconsistencia (*self-consistency*)** como una forma de enfatizar la necesidad de generar múltiples respuestas y pasos de razonamiento.

### Muestras *Best-of-N*

El primer método que involucra un verificador se llama **Muestras *Best-of-N***. Esta técnica genera $N$ muestras y luego utiliza un verificador (Modelo de Recompensa de Resultado, ORM) para juzgar cada respuesta:

Primero, el LLM (a menudo llamado el Proponente) genera múltiples respuestas utilizando una temperatura alta o variable.

[Image: Diagrama que ilustra la generación de múltiples respuestas (Best-of-N).]

Segundo, cada respuesta se pasa a través de un **Modelo de Recompensa de Resultado (ORM)** y se puntúa la calidad de la respuesta. Se selecciona la respuesta con la puntuación más alta:

[Image: Diagrama que ilustra la puntuación de las respuestas con un ORM.]

En lugar de juzgar la respuesta, el proceso de razonamiento también podría ser juzgado con un **Modelo de Recompensa de Proceso (PRM)** que juzga la calidad de cada paso de razonamiento. Elegirá el candidato con el peso total más alto.

[Image: Diagrama que ilustra la puntuación del proceso con un PRM.]

Con ambos tipos de verificadores, también podemos ponderar cada respuesta candidata por el RM y elegir aquella con el peso total más alto. Esto se llama **Muestras *Best-of-N* Ponderadas**:

[Image: Diagrama que ilustra el muestreo ponderado Best-of-N.]

---

## Búsqueda *Beam* con Modelos de Recompensa de Proceso

Este proceso de generación de respuestas y pasos intermedios puede extenderse aún más con la **búsqueda *beam*** (*beam search*). Con la búsqueda *beam*, se muestrean múltiples pasos de razonamiento y cada uno es juzgado por PRM (similar al Árbol de Pensamiento). Los 3 principales "**haces**" (*beams*) (caminos con mejor puntuación) son rastreados a lo largo del proceso.

[Image: Diagrama que ilustra la búsqueda *Beam* con un PRM.]

Este método permite detener rápidamente los caminos de "razonamiento" que no resultan fructíferos (puntuados bajos por el PRM).

Las respuestas resultantes se ponderan luego utilizando el enfoque *Best-of-N* que exploramos antes.

---

## Búsqueda en Árbol Monte Carlo (*Monte Carlo Tree Search*)

Una gran técnica para hacer que las búsquedas en árbol sean más eficientes se llama **Búsqueda en Árbol Monte Carlo (MCTS)**. Consta de cuatro pasos:

* Selección (seleccionar una hoja dada basándose en una fórmula predeterminada)
* Expansión (crear nodos adicionales)
* Despliegues (*Rollouts*) (crear nodos aleatoriamente hasta llegar al final)
* Retropropagación (*Backprop*) (actualizar las puntuaciones de los nodos padres basándose en la salida)

El objetivo principal de estos pasos es seguir expandiendo los mejores pasos de razonamiento mientras también se exploran otros caminos.

Es, por lo tanto, un equilibrio entre **exploración** y **explotación**. Un ejemplo de cómo se puntúan y seleccionan los nodos es el siguiente:

[Image: Fórmula utilizada para la selección y puntuación de nodos en MCTS (fórmula Upper Confidence Bound 1 - UCB1).]

Por lo tanto, cuando seleccionamos un nuevo paso de razonamiento para explorar, no tiene que ser el camino de mejor rendimiento hasta ahora.

Usando este tipo de fórmula, comenzamos **seleccionando** un nodo (paso de razonamiento) y lo **expandimos** generando nuevos pasos de razonamiento. Como antes, esto se puede hacer con valores de **temperatura** razonablemente altos y variables:

[Image: Diagrama de la fase de Selección y Expansión en MCTS.]

Uno de los pasos de razonamiento expandidos se selecciona y se despliega (*rolled out*) varias veces hasta que alcanza varias respuestas.

Estos despliegues se pueden juzgar basándose en los pasos de razonamiento (PRM), las recompensas (ORM) o una combinación de ambos.

Las puntuaciones de los nodos padres se actualizan (retropropagadas) y podemos comenzar el proceso de nuevo comenzando con la selección.

[Image: Diagrama que ilustra el ciclo completo de MCTS (Selección, Expansión, Despliegue y Retropropagación).]

## Modificación de la Distribución de Propuesta

La segunda categoría para crear LLMs de razonamiento se llama "**Modificación de la Distribución de Propuesta**". En lugar de buscar los pasos de razonamiento correctos con verificadores (**centrado en la salida**), el modelo es entrenado para crear pasos de razonamiento mejorados (**centrado en la entrada**).

En otras palabras, la distribución de la que se muestrean las terminaciones/pensamientos/tokens se modifica.

Imagine que tenemos una pregunta y una distribución de la que podemos muestrear *tokens*. Una estrategia común sería obtener el *token* con la puntuación más alta:

[Image: Diagrama que muestra la selección del *token* con la puntuación más alta (greedy decoding).]

Sin embargo, observe que algunos de los *tokens* están coloreados de rojo en la imagen de arriba. Esos son los *tokens* que tienen más probabilidades de conducir a un proceso de razonamiento:

[Image: Distribución de *tokens* donde algunos están coloreados de rojo (probables *tokens* de razonamiento).]

Aunque tomar el *token* ambicioso (*greedy*) no es necesariamente incorrecto, seleccionar un *token* que conduzca a un proceso de razonamiento tiende a resultar en una respuesta mejorada.

Cuando modificamos la distribución de propuesta (la distribución de probabilidad de *tokens*), esencialmente estamos haciendo que el modelo reordene la distribución de tal manera que los *tokens* de "razonamiento" se seleccionen con más frecuencia:

[Image: Diagrama que muestra la distribución de *tokens* reordenada para priorizar los *tokens* de razonamiento.]

Hay varios métodos para modificar la distribución de propuesta, pero generalmente se pueden clasificar en dos categorías:

* Actualizar el ***prompt*** a través de la ingeniería de *prompts*
* **Entrenar** el modelo para centrarse en los *tokens*/procesos de razonamiento

---

## Prompting

Con la ingeniería de *prompts*, intentamos mejorar la salida actualizando el *prompt*. Este proceso también puede empujar al modelo a mostrar algunos de los procesos de razonamiento que vimos antes.

Para cambiar la distribución de propuesta con *prompting*, podemos proporcionar ejemplos al modelo (*in-context learning*) que tiene que seguir para generar un comportamiento similar al razonamiento:

[Image: Diagrama que ilustra la ingeniería de *prompts* con ejemplos (CoT *Few-Shot*).]

Este proceso se puede simplificar aún más simplemente indicando "*Pensemos paso a paso*". Del mismo modo, esto cambia la distribución de propuesta de tal manera que el LLM tiende a descomponer el proceso antes de responder:

[Image: Diagrama que ilustra el CoT *Zero-Shot*.]

Sin embargo, el modelo no ha aprendido inherentemente a seguir este proceso. Además, este es un proceso estático y lineal que inhibe el auto-refinamiento. Si un modelo comienza con un proceso de razonamiento incorrecto, tiende a mantenerlo en lugar de revisarlo.

---

## STaR

Además del *prompting*, también podemos hacer que el modelo aprenda a "razonar" entrenándolo para que sea recompensado por generar estos pasos de razonamiento. Esto generalmente implica un conjunto de datos de razonamiento y aprendizaje por refuerzo para recompensar ciertos comportamientos.

Una técnica muy debatida se llama **STaR** o **Self-Taught Reasoner** (Razonador Auto-Enseñado). STaR es un método que utiliza el LLM para generar sus propios datos de razonamiento como entrada para el ajuste fino (*fine-tuning*) del modelo.

En el primer paso (1), genera pasos de razonamiento y una respuesta. Si la respuesta es correcta (2a), entonces añade el razonamiento y la respuesta a un conjunto de datos de entrenamiento en triplete (3b). Estos datos se utilizan para realizar el **ajuste fino supervisado** del modelo (5):

[Image: Diagrama del proceso STaR cuando la respuesta inicial es Correcta (pasos 1, 2a, 3b, 5).]

Sin embargo, si el modelo proporciona una respuesta incorrecta (2b), entonces proporcionamos una "pista" (la respuesta correcta) y le pedimos al modelo que razone por qué esta respuesta sería correcta (4b). El paso de razonamiento final se añade al mismo conjunto de datos de entrenamiento en triplete que se utiliza para realizar el **ajuste fino supervisado** del modelo (5):

[Image: Diagrama del proceso STaR cuando la respuesta inicial es Incorrecta (pasos 1, 2b, 4b, 5).]

Un elemento clave aquí (junto con muchas otras técnicas para modificar la distribución de propuesta) es que entrenamos explícitamente al modelo para que siga los procesos de razonamiento que le mostramos.

En otras palabras, decidimos cómo debe ser el proceso de razonamiento a través del **ajuste fino supervisado**.

El *pipeline* completo es bastante interesante ya que esencialmente genera **ejemplos de entrenamiento sintéticos**. Usar ejemplos de entrenamiento sintéticos (como exploraremos en DeepSeek R-1) es un método increíble para **destilar** este proceso de razonamiento en otros modelos.

## DeepSeek-R1

Un lanzamiento importante en modelos de razonamiento es **DeepSeek-R1**, un modelo de código abierto cuyos pesos están disponibles. Compitiendo directamente con el modelo de razonamiento OpenAI o1, DeepSeek-R1 ha tenido un gran impacto en el campo.

DeepSeek ha hecho un trabajo notable al destilar elegantemente el razonamiento en su modelo base (**DeepSeek-V3-Base**) a través de varias técnicas.

Curiosamente, no se utilizaron verificadores y, en lugar de utilizar el ajuste fino supervisado para destilar el comportamiento de razonamiento, un gran enfoque se centró en el **aprendizaje por refuerzo**.

¡Exploremos cómo entrenaron el comportamiento de razonamiento en sus modelos!

### Razonamiento con DeepSeek-R1 Zero

Un gran avance que condujo a DeepSeek-R1 fue un modelo experimental llamado **DeepSeek-R1 Zero**.

Comenzando con **DeepSeek-V3-Base**, en lugar de utilizar el ajuste fino supervisado en un montón de datos de razonamiento, solo utilizaron el **aprendizaje por refuerzo (RL)** para habilitar el comportamiento de razonamiento.

Para hacerlo, comenzaron con un *prompt* muy sencillo (similar a un *prompt* del sistema) para ser utilizado en el *pipeline*:

[Image: El *prompt* del sistema utilizado en el entrenamiento de DeepSeek-R1 Zero, que especifica el uso de *tags* &lt;think&gt; y &lt;answer&gt;.]

Observe cómo mencionan explícitamente que el proceso de razonamiento debe ir entre las ***tags* &lt;think&gt;**, ¡pero no especifican cómo debe ser el proceso de razonamiento!

Durante el aprendizaje por refuerzo, se crearon dos recompensas específicas basadas en reglas:

* Recompensas de Precisión (*Accuracy rewards*) - Recompensa la **respuesta** probándola.
* Recompensas de Formato (*Format rewards*) - Recompensa el uso de las **tags &lt;thinking&gt;** y **&lt;answer&gt;**.

El algoritmo de RL utilizado en este proceso se llama **Optimización de Política Relativa a Grupos (GRPO)**. La intuición detrás de este algoritmo es que hace que todas las elecciones que condujeron a una respuesta correcta o incorrecta sean más o menos probables. Estas elecciones pueden ser tanto conjuntos de *tokens* como pasos de razonamiento.

[Image: Diagrama que ilustra el proceso de GRPO (Muestreo, Recompensa de Grupo y Actualización de Política).]

Curiosamente, no se dieron ejemplos sobre cómo debería ser el proceso &lt;think&gt;. ¡Simplemente indica que **debe usar las *tags* &lt;think&gt;**, y nada más!

Al proporcionar estas recompensas indirectas relacionadas con el comportamiento de Cadena de Pensamiento, el modelo aprendió por sí mismo que cuanto más largo y complejo era el proceso de razonamiento, más probable era que la respuesta fuera correcta.

[Image: Gráfico que muestra la relación entre la longitud del razonamiento y la precisión.]

Este gráfico es especialmente importante ya que refuerza el **cambio de paradigma** del cómputo en **tiempo de entrenamiento** al cómputo en **tiempo de prueba**. A medida que estos modelos generan secuencias de pensamientos más largas, se centran en el cómputo en **tiempo de prueba**.

Utilizando este *pipeline* de entrenamiento, descubrieron que el modelo descubre por sí mismo el comportamiento más óptimo similar a la Cadena de Pensamiento, incluyendo capacidades de razonamiento avanzadas como la auto-reflexión y la auto-verificación.

Sin embargo, todavía tenía un inconveniente significativo. Tenía poca legibilidad y tendía a mezclar idiomas. En su lugar, exploraron una alternativa, el ahora conocido **DeepSeek R1**.

¡Exploremos cómo estabilizaron el proceso de razonamiento!

## DeepSeek-R1

Para crear DeepSeek-R1, los autores siguieron cinco pasos:

1.  Arranque en Frío (*Cold Start*)
2.  Aprendizaje por Refuerzo Orientado al Razonamiento
3.  Muestreo por Rechazo (*Rejection Sampling*)
4.  Ajuste Fino Supervisado
5.  Aprendizaje por Refuerzo para todos los Escenarios

En el **paso 1**, DeepSeek-V3-Base se ajustó con un pequeño conjunto de datos de razonamiento de alta calidad (≈5.000 *tokens*). Esto se hizo para prevenir el problema del arranque en frío que resultaba en una mala legibilidad.

[Image: Diagrama del Paso 1 (Ajuste Fino de Arranque en Frío).]

En el **paso 2**, el modelo resultante se entrenó utilizando un proceso de RL similar al utilizado para entrenar DeepSeek-R1-Zero. Sin embargo, se añadió otra medida de recompensa para asegurar que el idioma objetivo se mantuviera consistente.

[Image: Diagrama del Paso 2 (RL Orientado al Razonamiento con recompensa de idioma).]

En el **paso 3**, el modelo resultante entrenado con RL se utilizó para generar datos de razonamiento sintéticos que se utilizarían para el ajuste fino supervisado en una etapa posterior. Mediante el muestreo por rechazo (recompensas basadas en reglas) y un modelo de recompensa (DeepSeek-V3-Base), se crearon 600.000 muestras de razonamiento de alta calidad.

Además, se crearon 200.000 muestras sin razonamiento utilizando DeepSeek-V3 y parte de los datos con los que se entrenó.

[Image: Diagrama del Paso 3 (Generación de Muestras de Razonamiento Sintético).]

En el **paso 4**, el conjunto de datos resultante de 800.000 muestras se utilizó para realizar el ajuste fino supervisado del modelo DeepSeek-V3-Base.

[Image: Diagrama del Paso 4 (Ajuste Fino Supervisado con el conjunto de datos sintético).]

En el **paso 5**, se realizó el entrenamiento basado en RL en el modelo resultante utilizando un enfoque similar al utilizado en DeepSeek-R1-Zero. Sin embargo, para alinearse con las preferencias humanas, se agregaron señales de recompensa adicionales centradas en la utilidad y la inofensividad.

### Destilación del Razonamiento con DeepSeek-R1

DeepSeek-R1 es un modelo enorme con 671B de parámetros. Desafortunadamente, esto significa que ejecutar un modelo de este tipo será difícil en *hardware* de consumo.

Afortunadamente, los autores exploraron formas de destilar la calidad del razonamiento de DeepSeek-R1 en otros modelos, como Qwen-32B, ¡que podemos ejecutar en *hardware* de consumo!

Para hacerlo, utilizan DeepSeek-R1 como modelo **maestro** y el modelo más pequeño como **estudiante**. A ambos modelos se les presenta un *prompt* y tienen que generar una distribución de probabilidad de *tokens*. Durante el entrenamiento, el **estudiante** intentará seguir de cerca la distribución del **maestro**.

## Intentos Fallidos

¿Recuerda cuando hablamos de **Modelos de Recompensa de Proceso** (PRMs) y la **Búsqueda en Árbol Monte Carlo** (MCTS)? Resulta que DeepSeek también probó esas técnicas para inculcar el razonamiento, pero no tuvo éxito al hacerlo.

Con MCTS, encontraron problemas con el gran espacio de búsqueda y tuvieron que limitar las expansiones de nodos. Además, entrenar un Modelo de Recompensa de grano fino es intrínsecamente difícil.

Con los PRMs para las técnicas *Best-of-N*, encontraron problemas con la sobrecarga computacional para seguir reentrenando el Modelo de Recompensa para prevenir el *reward-hacking*.

¡Esto no significa que no sean técnicas válidas, pero proporciona información interesante sobre las limitaciones de estas técnicas!
