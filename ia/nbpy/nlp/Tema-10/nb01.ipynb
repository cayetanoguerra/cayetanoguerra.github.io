## **1. Arquitectura interna del Barcelona Supercomputing Center**

### **Naturaleza institucional**

El BSC-CNS es un consorcio público formado por:

* **Gobierno de España (Ministerio de Ciencia e Innovación)** – 60 %
* **Generalitat de Catalunya** – 30 %
* **Universitat Politècnica de Catalunya (UPC)** – 10 %

Su misión es doble:

1. **Investigar** en ciencias de la computación y en aplicaciones científicas.
2. **Proveer servicios de HPC** a la comunidad científica, gestionando, además, la Red Española de Supercomputación (RES).

### **Estructura científica y de servicios**

A nivel interno, el BSC se organiza en grandes áreas o departamentos de investigación y soporte:

* **Computer Sciences**
* **Life Sciences (Ciencias de la Vida)**
* **Earth Sciences (Ciencias de la Tierra)**
* **Computer Applications in Science and Engineering (Aplicaciones Computacionales en Ciencia e Ingeniería)**
* Más recientemente, se ha añadido un área de **Ciencias Sociales y Humanidades Computacionales**, reflejando el uso de HPC y datos masivos en estas disciplinas.

Además, existe una estructura de **Operaciones** (que incluye la gestión de MareNostrum y otros sistemas), **Transferencia tecnológica y spin-offs**, y gestión de la **RES** y de proyectos europeos EuroHPC.

## **2. Arquitectura y estructura de MareNostrum 5**

MareNostrum 5 (MN5) es un supercomputador **pre-exaescala** de la infraestructura EuroHPC, instalado en el BSC y operativo como uno de los sistemas más potentes de Europa.

### **Visión general**
* **Tipo de sistema**: supercomputador híbrido, orientado a HPC clásico y a IA/big data.
* **Proveedor / plataforma**: combina arquitecturas **Bull Sequana XH3000** y **Lenovo ThinkSystem**.
* **CPU principal**: **Intel Xeon “Sapphire Rapids”**.
* **GPU principal**: **NVIDIA Hopper** en la partición acelerada.
* **Sistema operativo**: familia **Red Hat Enterprise Linux**.

### **Particiones principales**

La arquitectura de MN5 se organiza en varias **particiones**, cada una optimizada para un tipo de carga. Las dos grandes son:

1. **GPP – General Purpose Partition (partición CPU)**

   * ~**6480 nodos** basados en **Intel Sapphire Rapids**.
   * Memoria DDR5 por nodo.
   * Orientada a **HPC clásico**: simulación numérica, CFD, dinámica de fluidos, modelos climáticos, mecánica estructural, etc.
   * Rinde del orden de **decenas de petaflops** (≈46 PFlops de pico para esta partición según documentación técnica reciente).

2. **ACC – Accelerated Partition (partición acelerada CPU+GPU)**

   * **1120 nodos** con **Intel Sapphire Rapids + GPUs NVIDIA Hopper**.
   * Combinación de **memoria DDR5** y **HBM** en las GPUs, para altas tasas de ancho de banda.
   * Especializada en:

     * **IA y deep learning** a gran escala.
     * Cálculos híbridos HPC+IA (p. ej. simulaciones aceleradas por redes neuronales).
     * Aplicaciones de dinámica molecular, bioinformática, etc., muy dependientes de GPU.


### **Red de interconexión y E/S**

* **Red de interconexión**: **InfiniBand NDR200** de alta velocidad, que da soporte a las comunicaciones MPI y a los accesos de E/S del sistema paralelo.
* **Almacenamiento**:

  * Más de **200 PB** de capacidad principal de almacenamiento y hasta ~**400 PB** de archivo activo, según EuroHPC.
  * Documentación reciente de la RES habla de del orden de **centenares de petabytes** de almacenamiento compartido entre particiones (por ejemplo, ~650 PB en ciertas descripciones de la partición ACC en la RES).


## **3. Características técnicas globales de MareNostrum 5**

  * Gemelos digitales de órganos y células.

  * Modelos climáticos de alta resolución.

  * Factorías europeas de IA.
  * Espacios de datos médicos y otros dominios científicos.

## **SLURM Workload Manager**
<div align="left">
    <img src="imgs/Slurm_logo.svg.png" width="250px">
</div>

SLURM (Slurm Workload Manager) es un **gestor de colas y planificador de recursos** para sistemas de cómputo de altas prestaciones (HPC). Se utiliza para **administrar el envío, planificación y ejecución de trabajos** en clusters con muchos nodos y usuarios concurrentes.

### **¿Qué es SLURM?**

* Es un **sistema de gestión de trabajos** (workload manager) y **planificador de recursos** (resource manager).
* Permite que múltiples usuarios compartan de forma ordenada un **cluster de cómputo**.
* Es **software libre** (licencia GPL) y está ampliamente utilizado en centros HPC de investigación, universidades y supercomputadores.

### **Funciones principales**

1. **Gestión de recursos**

   * Controla nodos, CPUs, memoria, GPUs, etc.
   * Agrupa recursos en **particiones** (queues) con distintas políticas (tiempo máximo, prioridad, tipo de nodo…).

2. **Gestión de trabajos (jobs)**

   * Permite enviar trabajos batch y jobs interactivos.
   * Asigna recursos a cada trabajo según los requisitos que indica el usuario.
   * Gestiona estados de los trabajos: *PENDING, RUNNING, COMPLETED, FAILED,* etc.

3. **Planificación (scheduling)**

   * Decide **cuándo** y **dónde** se ejecuta cada trabajo.
   * Tiene en cuenta prioridad, tiempo en cola, recursos solicitados, uso histórico del usuario, etc.
   * Puede aplicar políticas como *backfilling* (rellenar huecos con trabajos pequeños).

### **Componentes básicos**

* **slurmctld**: demonio controlador central. Mantiene el estado del cluster y planifica los trabajos.
* **slurmd**: demonio que corre en cada nodo de cómputo y ejecuta los trabajos.
* **slurmdbd** (opcional): demonio de base de datos para contabilidad (accounting).
* **Comandos de usuario** (cliente):

  * `sbatch`: enviar trabajos batch.
  * `srun`: ejecutar paralelamente (a veces dentro de un script de `sbatch`).
  * `squeue`: ver la cola de trabajos.
  * `scancel`: cancelar trabajos.
  * `sinfo`: ver el estado de nodos y particiones.

### **Flujo típico de uso**

1. El usuario escribe un **script de trabajo** (por ejemplo, bash) donde indica:

   * Recursos (nº de nodos/CPUs, tiempo, memoria, partición, etc.) mediante directivas `#SBATCH`.
   * Comandos a ejecutar (por ejemplo, ejecutar un programa paralelo con MPI).
2. Envía el trabajo con `sbatch script.sh`.
3. SLURM coloca el trabajo en la **cola** y espera a que haya recursos disponibles.
4. Cuando se asignan recursos, el trabajo pasa a *RUNNING* y se ejecuta en los nodos.
5. Al finalizar, genera ficheros de salida y el estado se marca como *COMPLETED* o *FAILED*.



# **Paralelización**

## **Paralelización de los datos**

Cuando se entrena un modelo grande (por ejemplo, un LLM) en un sistema HPC, la forma más sencilla y clásica de aprovechar varias GPUs es el **paralelismo de datos**. La idea básica es muy intuitiva: se mantiene **el mismo modelo completo** replicado en varias GPUs y lo que se reparte entre ellas son los **datos del mini-batch**.

Imaginemos un nodo con 4 GPUs (GPU0, GPU1, GPU2 y GPU3), unidas por una interconexión rápida tipo NVLink. El modelo está copiado íntegramente en cada una de las cuatro GPUs; es decir, cada GPU tiene su propia copia de todos los pesos. Supongamos que el tamaño de mini-batch global que queremos usar en el entrenamiento es de 64 ejemplos. En un esquema de paralelismo de datos, ese mini-batch global se divide automáticamente en cuatro mini-lotes locales: a cada GPU le corresponden 16 ejemplos. Desde el punto de vista de cada GPU, lo que ve es “un mini-batch de 16 ejemplos” y un modelo completo que puede usar para hacer forward y backward.

Durante el forward, cada GPU trabaja de manera independiente con sus 16 ejemplos. Calcula las activaciones capa a capa y, al final, la función de pérdida sobre esos datos locales. Normalmente, la pérdida que se maneja en cada GPU ya está promediada sobre sus 16 ejemplos (por ejemplo, la media de la cross-entropy de esos 16 inputs). A partir de esa pérdida local, se lanza el backward: cada GPU calcula los gradientes de todos los parámetros del modelo respecto a su pérdida local. El resultado es que, al terminar el backward, tenemos cuatro conjuntos de gradientes: ∇W₀ en la GPU0, ∇W₁ en la GPU1, ∇W₂ en la GPU2 y ∇W₃ en la GPU3. Esos gradientes tienen la misma forma (mismo número de tensores, mismas dimensiones), pero cada uno refleja únicamente la contribución de los 16 ejemplos que esa GPU ha visto.

Hasta aquí, cada réplica ha entrenado como si fuera un proceso independiente con su mini-batch. Para que el entrenamiento realmente se comporte como si estuviéramos usando un único modelo y un mini-batch de 64 ejemplos, es necesario **combinar** estos gradientes locales en un gradiente global. Esa combinación se hace mediante una operación colectiva de comunicación llamada **all-reduce**, implementada a nivel de librería (por ejemplo, NCCL en PyTorch DDP).

En el all-reduce, las cuatro GPUs se comunican entre sí para sumar (o promediar) sus gradientes de manera coordinada. Conceptualmente, lo que se hace es calcular

$$
\nabla W = \frac{1}{4} \big( \nabla W_0 + \nabla W_1 + \nabla W_2 + \nabla W_3 \big),
$$

es decir, el gradiente medio sobre los 64 ejemplos del mini-batch global. La operación se hace elemento a elemento sobre cada tensor de pesos, y se realiza de forma distribuida: los gradientes se “pasean” entre GPUs, se van sumando y al final **cada GPU termina teniendo una copia idéntica del gradiente global**. Hay variantes en las que se hace primero la suma y luego se divide por el número de réplicas, y variantes en las que se deja la suma y se ajusta el learning rate; desde el punto de vista conceptual, suma y media son equivalentes si se ajustan coherentemente los hiperparámetros.

Una vez que las cuatro GPUs disponen del mismo gradiente global, cada una aplica de manera local el paso del optimizador (SGD, Adam, AdamW, etc.) sobre su copia de los pesos:

$$
W \leftarrow W - \eta \nabla W.
$$

Dado que antes de este paso todas las copias de W eran iguales y el gradiente ∇W que usan también es el mismo en las cuatro GPUs, tras la actualización los pesos quedan de nuevo **sincronizados**: las cuatro réplicas del modelo tienen exactamente los mismos parámetros. Desde el punto de vista matemático, esto es equivalente a haber tenido un único modelo en una sola GPU procesando un mini-batch de 64 ejemplos.

## **Paralelización del modelo**

Cuando el tamaño del modelo crece hasta el punto de que **ya no cabe completo en la memoria de una sola GPU**, o cuando se quiere explotar mejor la interconexión rápida entre varias GPUs, se recurre al **paralelismo de modelo**. A diferencia del paralelismo de datos, aquí no se replican todos los parámetros en cada GPU; lo que se hace es **partir el propio modelo** y distribuir sus parámetros y cálculos entre varios dispositivos. El objetivo es que, colectivamente, varias GPUs puedan ejecutar un modelo que individualmente no podrían alojar.

Existen dos formas principales de paralelismo de modelo que se usan en el entrenamiento de LLMs: el **paralelismo tensorial (o intra-capa)** y el **paralelismo por pipeline (o inter-capas)**. En la práctica suelen combinarse, pero conviene entenderlas por separado.

### **Paralelismo tensorial**

En el **paralelismo tensorial**, una misma capa del modelo se divide entre varias GPUs. Esto es especialmente natural en capas con grandes multiplicaciones de matrices, como las capas lineales de un Transformer (proyecciones QKV, MLPs, etc.). Por ejemplo, una capa lineal con una matriz de pesos muy grande puede dividirse por columnas o por filas: cada GPU almacena solo una fracción de esos pesos y calcula su parte del resultado. Durante el forward, cada GPU recibe la activación de entrada completa, multiplica por su trozo de la matriz y produce una salida parcial. Esas salidas parciales se combinan mediante operaciones colectivas (típicamente all-reduce o all-gather) para obtener la salida completa de la capa. Desde el punto de vista matemático, el resultado es idéntico al de ejecutar la capa en una sola GPU, pero el almacenamiento y el cómputo se han repartido.

Este tipo de paralelismo es muy eficiente cuando las GPUs están fuertemente conectadas, por ejemplo dentro de un mismo nodo mediante NVLink, porque requiere **comunicación frecuente y de baja latencia** entre dispositivos. Se utiliza mucho en implementaciones como Megatron-LM, donde cada capa del Transformer está cuidadosamente diseñada para que la comunicación necesaria sea mínima y pueda solaparse con el cómputo. El gran beneficio del paralelismo tensorial es que permite ejecutar capas enormes con un alto grado de paralelismo, pero la contrapartida es que el modelo queda muy “acoplado” a la topología hardware: las GPUs que comparten una capa deben comunicarse constantemente.


#### **Ejemplo**

Vamos a ver este tipo de paralelización tensorial con un ejemplo sencillo:

Supongamos que tenemos una capa lineal muy sencilla, sin sesgo, que hace:

$$
y = W x
$$

Supongamos que:

* (x) es un vector de tamaño 4.
* (y) es un vector de tamaño 2.
* Por tanto, (W) es una matriz de tamaño ($2 \times 4$).

Es decir,

$$
W =
\begin{bmatrix}
w_{11} & w_{12} & w_{13} & w_{14} \\
w_{21} & w_{22} & w_{23} & w_{24}
\end{bmatrix}, \quad
x =
\begin{bmatrix}
x_1 \ x_2 \ x_3 \ x_4
\end{bmatrix}
$$

y la salida es

$$
y =
\begin{bmatrix}
y_1 \ y_2
\end{bmatrix}
=
W x
$$

Ahora imagina que queremos hacer **paralelismo tensorial en 2 GPUs** dividiendo la matriz de pesos **por columnas**. Lo que hacemos es partir (W) en dos bloques:

$$
W =
\Big[
\underbrace{
\begin{bmatrix}
w_{11} & w_{12} \\
w_{21} & w_{22}
\end{bmatrix}
}_{W^{(0)}}
\;\;
\underbrace{
\begin{bmatrix}
w_{13} & w_{14} \\
w_{23} & w_{24}
\end{bmatrix}
}_{W^{(1)}}
\Big]
$$


Partimos también el vector de entrada $x$ en dos trozos compatibles:

$$
x =
\begin{bmatrix}
x_1 \\
x_2 \\
x_3 \\
x_4
\end{bmatrix}
=
\begin{bmatrix}
x^{(0)} \\
x^{(1)}
\end{bmatrix},
\quad \text{con} \quad
x^{(0)} =
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix},
\qquad
x^{(1)} =
\begin{bmatrix}
x_3 \\
x_4
\end{bmatrix}.
$$


donde $x^{(0)}$ contiene las dos primeras componentes y $x^{(1)}$ las dos últimas.

En el **forward** con paralelismo tensorial haríamos lo siguiente:

* En la **GPU 0** almacenamos $W^{(0)}$. Esta GPU recibe el trozo de entrada $x^{(0)}$ (o bien la entrada completa y se queda solo con lo que necesita) y calcula

$$
y^{(0)} = W^{(0)} x^{(0)} =
\begin{bmatrix}
w_{11} & w_{12} \\
w_{21} & w_{22}
\end{bmatrix}
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}
$$


* En la **GPU 1** almacenamos $W^{(1)}$. Esta GPU procesa $x^{(1)}$ y calcula

$$
y^{(1)} = W^{(1)} x^{(1)} =
\begin{bmatrix}
w_{13} & w_{14} \\
w_{23} & w_{24}
\end{bmatrix}
\begin{bmatrix}
x_3 \\
x_4
\end{bmatrix}
$$


Fíjate que (y^{(0)}) y (y^{(1)}) son vectores de tamaño 2. Si desarrollas el producto matricial original (Wx), verás que:

$$
y =
\begin{bmatrix}
y_1 \\
y_2
\end{bmatrix}
=
\begin{bmatrix}
w_{11} x_1 + w_{12} x_2 + w_{13} x_3 + w_{14} x_4 \\
w_{21} x_1 + w_{22} x_2 + w_{23} x_3 + w_{24} x_4
\end{bmatrix}
$$


Pero eso mismo se puede escribir como

$$
y = y^{(0)} + y^{(1)}
$$


porque

$$
y^{(0)} =
\begin{bmatrix}
w_{11} x_1 + w_{12} x_2 \\
w_{21} x_1 + w_{22} x_2
\end{bmatrix},
\qquad
y^{(1)} =
\begin{bmatrix}
w_{13} x_3 + w_{14} x_4 \\
w_{23} x_3 + w_{24} x_4
\end{bmatrix}
$$


Así que el **paso final del forward** en este paralelismo tensorial es una operación de comunicación muy sencilla: las dos GPUs se envían sus resultados parciales y se suman componente a componente. Cada GPU puede quedarse con la suma final o sólo una de ellas puede devolver el resultado, según cómo esté montado el framework, pero conceptualmente:

1. GPU 0 calcula $y^{(0)}$.
2. GPU 1 calcula $y^{(1)}$.
3. Se hace una suma distribuida ($y = y^{(0)} + y^{(1)}$) (por ejemplo, con un all-reduce sum).

En el **backward** ocurre algo análogo: cada GPU recibe el gradiente de la salida ($\partial L / \partial y$) (o su parte relevante) y, usando su trozo de la matriz de pesos, calcula sus gradientes locales
* ($\partial L / \partial W^{(0)}$) y ($\partial L / \partial x^{(0)}$) en la GPU 0,
* ($\partial L / \partial W^{(1)}$) y ($\partial L / \partial x^{(1)}$) en la GPU 1,

y después se combinan los gradientes sobre (x) de forma adecuada (de nuevo sumando componentes) para reconstruir ($\partial L / \partial x$).


### **Paralelismo por pipeline**

El **paralelismo por pipeline** aborda el problema de una forma distinta. En lugar de dividir cada capa, se divide el modelo en **bloques de capas consecutivas**, y cada bloque se asigna a una GPU o a un grupo de GPUs. Por ejemplo, en un Transformer con 48 capas, las capas 0–11 podrían ejecutarse en la primera GPU, las capas 12–23 en la segunda, las 24–35 en la tercera y las 36–47 en la cuarta. Durante el forward, los datos fluyen secuencialmente por esta “tubería”: la salida del primer bloque se envía al segundo, la del segundo al tercero, y así sucesivamente. En el backward ocurre el flujo inverso de gradientes.

Para que las GPUs no queden ociosas la mayor parte del tiempo, el mini-batch se divide en **micro-batches**. Mientras la primera GPU está procesando el forward del micro-batch 2, la segunda puede estar procesando el forward del micro-batch 1, y así sucesivamente. Este solapamiento permite aprovechar mejor los recursos, aunque siempre existen las llamadas “burbujas de pipeline” al inicio y al final de cada iteración, cuando algunas GPUs están inevitablemente esperando. La eficiencia del pipeline mejora cuanto mayor es el número de micro-batches, pero a cambio aumenta la complejidad y la latencia de una iteración completa.

Desde el punto de vista conceptual, el paralelismo por pipeline reduce el requisito de memoria en cada GPU, porque cada una solo necesita almacenar los parámetros y activaciones de sus propias capas. Sin embargo, introduce dependencias temporales claras entre las GPUs y una latencia estructural: los datos deben recorrer todo el pipeline para completar un forward/backward completo.

En la práctica moderna, el entrenamiento de LLMs grandes no utiliza una sola de estas técnicas de manera aislada. Lo habitual es combinar **paralelismo tensorial dentro de un nodo** (dividiendo las capas grandes entre varias GPUs muy próximas) con **paralelismo por pipeline entre grupos de GPUs o nodos**, y encima de todo aplicar **paralelismo de datos** para escalar a muchos nodos. A esto se le suele llamar paralelismo “3D”: datos, tensor y pipeline.
